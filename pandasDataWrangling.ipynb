```python
# Data wrangling with pandas 
# Creating a dataframe 
import pandas as pd 

data = {
    "name" : ["Aiyana", "Minho", "Zainab"],
    "age" : [21, 22, 20],
    "major" : ["CS", "Stats", "CS"],
    "gpa" : [3.85, 3.40, 3.92],
    "gender" : ["F", "M", "F"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Handling Missing Data 
# Detect missing values 
print(df.isna().sum())

# drop rows with missing values 
clean = df.dropna()

# fill missing GPA with the column mean 
df["gpa"] = df["gpa"].fillna(df["gpa"].mean())

# fill missing major with a placeholder 
df["major"] = df["major"].fillna("Unknown")
```
