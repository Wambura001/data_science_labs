```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Generate 'x' values (e.g., advertisement spend)
X = 2 * np.random.rand(100, 1)

# Generate 'y' values (e.g., sales) with a linear relationship and noise
y = 4 + 3 * X + np.random.randn(100, 1)

# Create a Pandas DataFrame for easier handling and viewing
df = pd.DataFrame({'Advertisement Spend (X)': X.flatten(), 'Sales (y)': y.flatten()})

print("First 5 rows of the generated data:")
display(df.head())
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Advertisement Spend (X)'], df['Sales (y)'], alpha=0.7)
plt.title('Synthetic Data: Advertisement Spend vs. Sales')
plt.xlabel('Advertisement Spend (X)')
plt.ylabel('Sales (y)')
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Create a Linear Regression model
lin_reg = LinearRegression()

# Fit the model to the data
lin_reg.fit(X, y)

# Make predictions using the trained model
y_pred = lin_reg.predict(X)

print("Model training complete.")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Advertisement Spend (X)'], df['Sales (y)'], alpha=0.7, label='Actual Data')
plt.plot(df['Advertisement Spend (X)'], y_pred, color='red', label='Regression Line')
plt.title('Linear Regression: Advertisement Spend vs. Predicted Sales')
plt.xlabel('Advertisement Spend (X)')
plt.ylabel('Sales (y)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
intercept = lin_reg.intercept_[0]
coefficient = lin_reg.coef_[0][0]
r_squared = r2_score(y, y_pred)

print(f"Intercept (Sales when Advertisement Spend is 0): {intercept:.2f}")
print(f"Coefficient (Impact of Advertisement Spend on Sales): {coefficient:.2f}")
print(f"R-squared (Model Fit): {r_squared:.2f}")
```
